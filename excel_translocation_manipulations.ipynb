```python
import pandas as pd
import os

# Set the directory where the CSV files are located
# directory = '/path/to/directory'

# Initialize an empty list to store the dataframes
df_list = []
num_columns_to_delete = 5

# Loop through all the files in the directory
for file in os.listdir(directory_save):
    # Skip files that are not CSV files
    if file.endswith('.csv'):
        # Load the CSV file into a dataframe
        df = pd.read_csv(os.path.join(directory_save, file))
        df.drop(df.columns[:num_columns_to_delete], axis=1, inplace=True)
        df = df.drop(df.index[:1])
        # Add the dataframe to the list
        df_list.append(df)

# Concatenate all the dataframes into a single dataframe
df_combined = pd.concat(df_list, axis=1)

# Write the combined dataframe to a new CSV file
df_combined.to_csv(directory_save+'combined_750uM.csv', index=False)
#df_combined.to_csv('F:/Sushil/DNAb/tAU/3252023.3/combined.csv', index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

# Load the CSV file
file_path = r'F:\Sushil\mcm\CMG\dNTP\111423 CMG MCM10 ssdna 1mm dntp\combined_750uM.csv'
df = pd.read_csv(file_path)

# Replace empty strings with NaN
df.replace(' ', np.nan, inplace=True)

# Extract x and y columns
x_cols = df.columns[:2]
y_cols = df.columns[2:]

# Plot all x-y pairs
plt.figure(figsize=(12, 6))
for i in range(0, len(y_cols), 2):
    x_data = df[x_cols[0]].astype(float)
    y_data = df[y_cols[i]].astype(float)

    # Drop NaN values
    mask = ~np.isnan(x_data) & ~np.isnan(y_data)
    x_data = x_data[mask]
    y_data = y_data[mask]

    plt.plot(x_data, y_data, label=f'Trace {i//2 + 1}')

    # Fit a linear line
    X = np.array(x_data).reshape(-1, 1)
    y = np.array(y_data)

    if len(X) > 1:
        model = LinearRegression().fit(X, y)

        # Check translocation rate and set color
        translocation_rate = model.coef_[0]
        if translocation_rate < 2:
            color = 'grey'
        else:
            color = plt.cm.viridis(i / len(y_cols))

        # Plot linear line
        plt.plot(X, model.predict(X), color=color, linestyle='--')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('X vs Y with Linear Fit')
plt.legend()
plt.show()

```
